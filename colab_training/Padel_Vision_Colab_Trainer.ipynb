## 3. Train YOLOv8 Padel Ball Detector on GPU
Fine-tune pre-trained YOLO on high-speed padel ball bounding boxes.

```python
from ultralytics import YOLO

# Load pre-trained lightweight YOLOv8s
ball_model = YOLO('yolov8s.pt')

# Sample training command (point 'data' to your padel_ball.yaml)
# ball_model.train(
#     data='padel_ball.yaml',
#     epochs=30,
#     imgsz=1080,
#     batch=16,
#     device=0,
#     name='padel_ball_run'
# )
print('YOLOv8 Ball model initialized and ready for training.')
```

## 4. Train YOLOv8 17-Keypoint Player Pose Estimator on GPU

```python
# Load pre-trained YOLOv8-pose model
pose_model = YOLO('yolov8s-pose.pt')

# Sample pose training command
# pose_model.train(
#     data='padel_pose.yaml',
#     epochs=35,
#     imgsz=960,
#     batch=16,
#     device=0,
#     name='padel_pose_run'
# )
print('YOLOv8 Pose model initialized and ready for training.')
```

## 5. Export Models to High-Speed ONNX (for Laptop CPU Inference)
ONNX models run at 4x speed on Intel laptop CPUs without requiring CUDA!

```python
# Export trained weights to ONNX format
onnx_ball = ball_model.export(format='onnx', dynamic=True, simplify=True)
onnx_pose = pose_model.export(format='onnx', dynamic=True, simplify=True)

print('Exported ONNX ball model:', onnx_ball)
print('Exported ONNX pose model:', onnx_pose)
```

## 6. Download Weights to Your Local Machine
Download the files and move them into your local `c:\Padel\models\` directory.

```python
from google.colab import files

try:
    files.download(onnx_ball)
    files.download(onnx_pose)
    print('Download started! Save files into c:\\Padel\\models\\')
except Exception as e:
    print('Download helper:', e)
    print('You can also download directly from the Colab file explorer panel on the left.')
```
